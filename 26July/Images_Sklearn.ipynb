```python
from sklearn.datasets import load_digits
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
load_digits().keys()
```

```python
dg = load_digits(n_class=3)
X = dg.images
Y = dg.target
```

```python
X.shape
```

```python
X[0]
```

```python
Y
```

```python
X.reshape(537, 64)
```

```python
Xflat = X.reshape(537, 64)
```

```python
from sklearn.neural_network import MLPClassifier
```

```python
model = MLPClassifier((128, 100, 50))
model.fit(Xflat, Y)
```

```python
Yp = model.predict(Xflat)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(Y, Yp)
```
