```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_excel('cluster.xlsx')
```

```python
X = df.values
```

```python
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score
```

```python
model = DBSCAN(eps = 0.5, min_samples = 4)
model.fit(X)
```

```python
silhouette_score(X, model.labels_)
```

```python
min_samples_ls = [3, 4, 5]
eps_ls = [1.5, 2, 2.5]
```

```python
for min_samples in min_samples_ls:
    for eps in eps_ls:
        # print(min_samples, eps)
        model = DBSCAN(eps = eps, min_samples = min_samples)
        model.fit(X)
        print(min_samples, eps, "===>", silhouette_score(X, model.labels_))
```

```python
mod = DBSCAN(min_samples=5, eps=2)
mod.fit(X)
mod.labels_
```

```python
plt.scatter(X[:, 0], X[:, 1], c=mod.labels_)
```
