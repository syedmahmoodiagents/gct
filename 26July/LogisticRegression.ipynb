```python
import mglearn as mg
```

```python
import pandas as pd
import numpy as np
```

```python
dftrain = pd.read_excel('clstrain.xlsx')
dftest = pd.read_excel('clstest.xlsx')
```

```python
x1 = dftrain['x1'].values
x2 = dftrain['x2'].values
t = dftrain['t']
```

```python
X = dftrain[['x1', 'x2']].values
y = dftrain['t'].values
```

```python
mg.discrete_scatter(x1, x2, t)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
model.fit(X, y)
```

```python
mg.discrete_scatter(x1, x2, t)
mg.plots.plot_2d_separator(model, X)
```

```python
model.predict_proba(X)
```

```python
np.argmax(model.predict_proba(X), axis=1)
```

```python
model.predict(X)
```

# Lets prepare the moon based dataset

```python
from sklearn.datasets import make_moons
```

```python
xx, tt = make_moons(n_samples=200 ,noise=0.15, random_state=120)
```

```python
xm1, xm2  = xx[:,0], xx[:,1]
```

```python
mg.discrete_scatter(xm1, xm2, tt)
```

```python
from sklearn.model_selection import train_test_split
```

```python
xxtrain, xxtest, yytrain, yytest = train_test_split(xx, tt)
```

```python
# yytrain
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
modtr = DecisionTreeClassifier(max_depth=4)
modtr.fit(xxtrain, yytrain)
```

```python
mg.discrete_scatter(xxtrain[:,0], xxtrain[:,1], yytrain)
mg.plots.plot_2d_separator(modtr, xxtrain)
```

# Logistic Regression

```python
modlog = LogisticRegression()
modlog.fit(xxtrain, yytrain)
```

```python
mg.discrete_scatter(xxtrain[:,0], xxtrain[:,1], yytrain)
mg.plots.plot_2d_separator(modlog, xxtrain)
```

Lets try with KNN
