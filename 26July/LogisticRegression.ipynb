```python
import mglearn as mg
```

```python
import pandas as pd
import numpy as np
```

```python
dftrain = pd.read_excel('clstrain.xlsx')
dftest = pd.read_excel('clstest.xlsx')
```

```python
x1 = dftrain['x1'].values
x2 = dftrain['x2'].values
t = dftrain['t']
```

```python
X = dftrain[['x1', 'x2']].values
y = dftrain['t'].values
```

```python
mg.discrete_scatter(x1, x2, t)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
model.fit(X, y)
```

```python
mg.discrete_scatter(x1, x2, t)
mg.plots.plot_2d_separator(model, X)
```

```python
model.predict_proba(X)
```

```python
np.argmax(model.predict_proba(X), axis=1)
```

```python
model.predict(X)
```

```python
x1_test = dftest['x1'].values
x2_test = dftest['x2'].values
t_test = dftest['t'].values
X_test = dftest[['x1', 'x2']]
```

```python
mg.discrete_scatter(x1_test, x2_test, t_test)
mg.plots.plot_2d_separator(model, X)
```

```python
from sklearn.metrics import accuracy_score
```

```python
yp = model.predict(X)
```

```python
accuracy_score(y, yp)
```

```python
yp_test = model.predict(X_test)
```

```python
accuracy_score(t_test, yp_test)
```
