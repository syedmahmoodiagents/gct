```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("blobdata.xlsx", index_col="Unnamed: 0")
```

```python
x = df[['x']].values
y = df['y'].values
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
mdtree = DecisionTreeRegressor(max_depth=3)
mdtree.fit(x,y)
```

```python
yptree = mdtree.predict(x)
```

```python
plt.scatter(x,y)
plt.scatter(x,yptree)
```

# Random_Forest with same max_depth

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
mdforest = RandomForestRegressor(max_depth=3)
mdforest.fit(x,y)
```

```python
ypforest = mdforest.predict(x)
```

```python
plt.scatter(x, y)
plt.scatter(x, ypforest)
```
