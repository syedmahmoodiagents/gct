```python
from sklearn.model_selection import train_test_split
```

```python
# train_test_split(df, random_state=0)
```

```python
x = df[['x']].values
y = df['y'].values
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(x, y, random_state=0)
```

```python
xtrain.shape
```

```python
from sklearn.metrics import mean_squared_error
```

```python
model1 = DecisionTreeRegressor(max_depth=1)
model1.fit(xtrain, ytrain)
```

```python
yptrain1 = model1.predict(xtrain)
mean_squared_error(ytrain, yptrain1)
```

```python
yptest1 = model1.predict(xtest)
mean_squared_error(ytest, yptest1)
```

```python
model2 = DecisionTreeRegressor(max_depth=2)
model2.fit(xtrain, ytrain)
```

```python
yptrain2 = model2.predict(xtrain)
mean_squared_error(ytrain, yptrain2)
```

```python
yptest2 = model2.predict(xtest)
mean_squared_error(ytest, yptest2)
```

```python
model3 = DecisionTreeRegressor(max_depth=3)
model3.fit(xtrain, ytrain)
```

```python
yptrain3 = model3.predict(xtrain)
mean_squared_error(ytrain, yptrain3)
```

```python
yptest3 = model3.predict(xtest)
mean_squared_error(ytest, yptest3)
```
