```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dftrain = pd.read_excel("train.xlsx")
dftest = pd.read_excel("test.xlsx")
```

```python
xtrain = dftrain[['xtrain']].values
ytrain = dftrain['ytrain'].values
```

```python
xtest = dftest[['xtest']].values
ytest = dftest['ytest'].values
```

```python
plt.scatter(xtrain, ytrain)
```

```python
from sklearn.linear_model import LinearRegression, Lasso, Ridge
from sklearn.metrics import mean_squared_error, r2_score
```

```python
mdreg = LinearRegression()
mdreg.fit(xtrain, ytrain)
yprg_tr = mdreg.predict(xtrain)
yprg_ts = mdreg.predict(xtest)
print(mean_squared_error(ytrain, yprg_tr))
print(mean_squared_error(ytest, yprg_ts))
```

```python
plt.scatter(xtrain, ytrain)
plt.plot(xtrain, yprg_tr)
```

# Regularization

```python
mdridge = Ridge(alpha=0)
mdridge.fit(xtrain, ytrain)
ypridge_tr = mdridge.predict(xtrain)
ypridge_ts = mdridge.predict(xtest)
print(mean_squared_error(ytrain, ypridge_tr))
print(mean_squared_error(ytest, ypridge_ts))
```

```python
plt.scatter(xtrain, ytrain)
plt.plot(xtrain, ypridge_tr)
```

```python
mdridge10 = Ridge(alpha=100)
mdridge10.fit(xtrain, ytrain)
ypridge10_tr = mdridge10.predict(xtrain)
ypridge10_ts = mdridge10.predict(xtest)
print(mean_squared_error(ytrain, ypridge10_tr))
print(mean_squared_error(ytest, ypridge10_ts))
```

```python
plt.scatter(xtrain, ytrain)
plt.plot(xtrain, ypridge10_tr)
```
